# **# BUSCANDO DADOS DOS PAPÉIS NO YFINANCE**

```python
DADOS_FILTRADOS_GITHUB_SA = ArqLer_Arq_0_1_1_Dados_Filtrados_Github['CODIGO'] + '.SA'
DADOS_FILTRADOS_GITHUB_SA
```

```python
DADOS_FILTRADOS_GITHUB_SA_Close = pd.DataFrame()

for x in DADOS_FILTRADOS_GITHUB_SA:
    try:
        data = yf.download(x, start=start, end=end, progress=False)
        if not data.empty and 'Close' in data.columns:
            DADOS_FILTRADOS_GITHUB_SA_Close[x] = data['Close']
        else:
            print(f'Aviso: Dados não encontrados para {x}')
    except Exception as e:
        print(f'Erro ao baixar {x}: {e}')

# Filtrando colunas que são todas zero ou todas vazias (NaN)
if not DADOS_FILTRADOS_GITHUB_SA_Close.empty:
    # Remove colunas que são apenas zero
    DADOS_FILTRADOS_GITHUB_SA_Close = DADOS_FILTRADOS_GITHUB_SA_Close.loc[:, (DADOS_FILTRADOS_GITHUB_SA_Close != 0).any(axis=0)]
    # Remove colunas que são totalmente vazias
    DADOS_FILTRADOS_GITHUB_SA_Close = DADOS_FILTRADOS_GITHUB_SA_Close.dropna(axis=1, how='all')
```

```python
DADOS_FILTRADOS_GITHUB_SA_Close
```

```python
# Obtém os nomes das colunas (tickers) que sobraram após a filtragem
tickers_existentes = DADOS_FILTRADOS_GITHUB_SA_Close.columns.tolist()

# Remove o sufixo '.SA' para voltar ao formato original
codigos_limpos = [ticker.replace('.SA', '') for ticker in tickers_existentes]

# Cria o novo DataFrame apenas com a coluna CODIGO
df_codigos_existentes = pd.DataFrame(codigos_limpos, columns=['CODIGO'])

# Exibe o resultado e a quantidade de papéis encontrados
print(f'Total de papéis com dados encontrados: {len(df_codigos_existentes)}')
```

```python
DADOS_GITHUB_EXISTENTES = df_codigos_existentes
DADOS_GITHUB_EXISTENTES
```
