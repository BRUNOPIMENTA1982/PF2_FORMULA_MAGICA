# **# DETERMINANDO PERÍODO DE CONSULTA PARA DADOS**

```python
start = dt.datetime(2011,1,1)     # VARIÁVEL START - DEFINIÇÃO DO PERÍODOD DE CONSULTA
end = dt.datetime(2019, 12, 31)     # VARIÁVEL END
```

# **# BAIXANDO DADOS DE FECHAMENTO DO YFINANCE**

```python
CARTEIRA_FM_SA = CARTEIRA_FM[['CODIGO']] + '.SA'
CARTEIRA_FM_SA
```

```python
CARTEIRA_FM_SA_CLOSE = pd.DataFrame()

# Corrigido: iterando sobre os valores da coluna 'CODIGO'
for x in CARTEIRA_FM_SA['CODIGO']:
    try:
        print(f'Baixando dados para: {x}...')
        data = yf.download(x, start=start, end=end, progress=False)
        if not data.empty and 'Close' in data.columns:
            # Garantindo que pegamos apenas a série de preços
            CARTEIRA_FM_SA_CLOSE[x] = data['Close']
        else:
            print(f'Aviso: Dados não encontrados para {x}')
    except Exception as e:
        print(f'Erro ao baixar {x}: {e}')

# Filtrando colunas que são todas zero ou todas vazias (NaN)
if not CARTEIRA_FM_SA_CLOSE.empty:
    CARTEIRA_FM_SA_CLOSE = CARTEIRA_FM_SA_CLOSE.loc[:, (CARTEIRA_FM_SA_CLOSE != 0).any(axis=0)]
    CARTEIRA_FM_SA_CLOSE = CARTEIRA_FM_SA_CLOSE.dropna(axis=1, how='all')
```

```python
CARTEIRA_FM_SA_CLOSE
```
