```python
DADOS_PAPEIS = ArqLer_DADOS_GITHUB_FECHADOS[['CODIGO', 'ROIC', 'EV/EBIT', 'SETOR']]
DADOS_PAPEIS
```

# **# ORDENANDO EV/EBIT**

```python
EVOrdenado = DADOS_PAPEIS.sort_values(by = 'EV/EBIT')     # CRIAÇÃO DA VARIÁVEL PARA ORDENAR EV/EBIT DO MAIOR PARA O MENOR
EVOrdenado = EVOrdenado.reset_index(drop = True)
EVOrdenado['EVOrdenado'] = EVOrdenado.index
EVOrdenado
```

# **# ORDENANDO ROIC**

```python
RoicOrdenado = EVOrdenado.sort_values(by = 'ROIC', ascending=False)      # CRIAÇÃO DA VARIÁVEL PARA ORDENAR ROIC DO MENOR PARA O MAIOR
RoicOrdenado = RoicOrdenado.reset_index(drop = True)
RoicOrdenado['RoicOrdenado'] = RoicOrdenado.index
RoicOrdenado
```

# **# RANKING DOS DADOS**

```python
Dados_Ordenados = RoicOrdenado

Dados_Ordenados['ranking'] = Dados_Ordenados['EVOrdenado'] + Dados_Ordenados['RoicOrdenado']     # ATRIBUINDO RANKING A VARIÁVEL DADOS
Dados_Ordenados.sort_values(by = 'ranking')
```

# **# FILTRANDO SETORES DE ATUAÇÃO**

```python
Dados_Ordenados['SETOR'].unique()     # APRESENTAÇÃO DE TODOS OS SETORES LISTADOS
```

```python


FiltroSetores = Dados_Ordenados[Dados_Ordenados['SETOR'] != 'Financeiro']     # RETIRADO OS SETORES CONFORME FÓRMULA
FiltroSetores = FiltroSetores[FiltroSetores['SETOR'] != 'Energia']
FiltroSetores = FiltroSetores[FiltroSetores['SETOR'] != 'Utilidade pública']
FiltroSetores = FiltroSetores[FiltroSetores['SETOR'] != 'NA']

FiltroSetores
```

# **# DETERMINANDO PAPÉIS PARA CARTEIRA FÓRMULA MÁGICA**

```python
CARTEIRA_FM = FiltroSetores.sort_values(by = 'ranking').head(30)  
CARTEIRA_FM
```
