```python
Dados_Papeis = ArqLer_Dados_Papeis[['CODIGO', 'Data_Registro_CVM', 'Situacao_Registro_CVM', 'Patrimônio Líquido Consolidado', 'trimestre', 'ano', 'ROIC', 'EV/EBIT', 'SETOR']]     # CRIAÇÃO DE UMA NOVA VARIÁVEL APENAS COM AS INFORMAÇÕES NECESSSÁRIAS PARA A FÓRMULA
Dados_Papeis


```

```python
Dados_Papeis = Dados_Papeis.fillna(0)     # SUBSTITUINDO VALORES NULOS POR 0 (ZERO) PARA EVITAR PROBLEMAS NA HORA DE APLICAR A FÓRMULA
Dados_Papeis
```

# **# FILTRANDO POR ANO DE REGISTRO NA CVM**

```python

Dados_Papeis_Filtro_Ano = Dados_Papeis[ Dados_Papeis['Data_Registro_CVM'].str[:4].astype(int) < 2011 ]
Dados_Papeis_Filtro_Ano
```

# **# FILTRANDO POR SITUAÇÃO**

```python

Dados_Papeis_Filtro_Situacao = Dados_Papeis_Filtro_Ano[ Dados_Papeis_Filtro_Ano['Situacao_Registro_CVM'] == "Ativo" ]
Dados_Papeis_Filtro_Situacao
```

# **# FILTRANDO POR PATRIMÔNIO LÍQUIDO POSITIVO**

```python
Dados_Papeis_Filtro_Patrimonio = Dados_Papeis_Filtro_Situacao[Dados_Papeis_Filtro_Situacao['Patrimônio Líquido Consolidado'] > 0]
Dados_Papeis_Filtro_Patrimonio
```

# **# FILTRANDO POR ANO DOS DADOS**

```python

Dados_Papeis_AnoDados = Dados_Papeis_Filtro_Patrimonio[Dados_Papeis_Filtro_Patrimonio['ano'].astype(int) == 2010]
Dados_Papeis_AnoDados
```

```python
Dados_Papeis_AnoDado_Trim = Dados_Papeis_AnoDados[Dados_Papeis_AnoDados['trimestre'] == 'annual']
Dados_Papeis_AnoDado_Trim
```

# **# CRIANDO ARQUIVO EXCEL PARA CARTEIRA FM**

```python
Dados_Filtrados_Github = Dados_Papeis_AnoDado_Trim
Dados_Filtrados_Github
```
